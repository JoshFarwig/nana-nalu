```python
import xarray as xr
import numpy as np

# NWPS model run from site: HFO (Hawaii), CG: 4 (Maui)
# download an example at: https://nomads.ncep.noaa.gov/gribfilter.php?ds=prnwps
# and set the following filename below to the downloaded one. place the downloaded 
# file into the notebooks/ dir for this ipynb to access it.

# NOTE: idx file created by grib engine or pygrib to improve read speed, 
# so when we need to refresh the data from the next model run, remove idx as well.

maui_grib2 = 'hfo_nwps_CG4_20251204_0000.grib2'
```

```python
an_ds = xr.open_dataset(
    maui_grib2,
    engine="cfgrib",
    filter_by_keys={'dataType': 'an'}
) 

fc_ds = xr.open_dataset(
    maui_grib2,
    engine="cfgrib",
    filter_by_keys={'dataType': 'fc'}
) 
```

# overall summary notes 

## x-array vs pygrib

x-array lazy loading is clutch, should reduce memory load and still allow me to search and pull specific 
values from lat / long w/ nearest neighbor selection. nearest neighbor probably makes the most sense, 
since bathymetry alongside coastline can change signifigantly on the islands, so bilinear interpolation 
could downplay any drasticd changes in bathymetry for nearshore waves. 

pygrib shines when you need to observe messaging format or write to grib files, which neither 
applies to the usecase intended, so x-array it is. 

## which data variables to use 

honestly, all data variables are useful. the two vars I considered scrapping were 
- dirc: current direction
- spc: speed of current

but these could be niche values not encompassed by most surf forecasting applications, so I think its valuable.
it's a safety precaution too, and can tell how hard you will be working on paddling back out hahaha.

### which swell height to portray... 

- shts: significant height of combined wind waves and swell
- swh: significant height of total swell

both are valuable. shts shows most accureate projection of nearshore wave height, including wind waves and storm surges. but doesn't nessicarily mean its "good", choppy wind waves aren't clean. if swh and shts are very similar to eachother, low wind chop and cleaner waves. both are important to show, just provide context on each one.  

```python
lat_spacing = fc_ds.latitude.diff("latitude").mean().values 
lon_spacing = fc_ds.longitude.diff("longitude").mean().values  

center_lat = fc_ds.latitude.mean().values 

# 1 degree lat ~ 111km everywhere 
lat_resolution = lat_spacing * 111000 

# 1 degree lon varies with latitude ~ 111km * cos(latitude)
lon_resolution = lon_spacing * 111000 * np.cos(np.radians(center_lat))

print(f"Latitude spacing: {lat_spacing:.6f}°")
print(f"Longitude spacing: {lon_spacing:.6f}°")
print(f"Cell resolution (lat): ~{lat_resolution:.0f} m")
print(f"Cell resolution (lon): ~{lon_resolution:.0f} m")
print(f"Approximate cell size: {lat_resolution:.0f}m x {lon_resolution:.0f}m")
```
